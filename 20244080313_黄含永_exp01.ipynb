```python

def score_level(score):
    if score < 0 or score > 100: raise ValueError("超出范围")
    if score >= 90: return "优秀"
    elif score >= 80: return "良好"
    elif score >= 60: return "及格"
    else: return "不及格"


def fact_rec(n):
    return 1 if n == 0 else n * fact_rec(n-1)
def fact_iter(n):
    res = 1
    for i in range(2, n+1): res *= i
    return res
```

```python
import os


data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)


content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
file_path = os.path.join(data_dir, "students.txt")
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)


def parse_student(line):
    parts = line.split(",")
    if len(parts) != 2: raise ValueError("字段数量错误")
    name, score_str = parts[0].strip(), parts[1].strip()
    if not name: raise ValueError("姓名为空")
    score = int(score_str) # 非数字抛错
    if not (0 <= score <= 100): raise ValueError("成绩超范围")
    return {"name": name, "score": score}


valid_records, errors = [], []
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()
    
for idx, line in enumerate(lines[1:], start=2): # 跳过表头
    try:
        valid_records.append(parse_student(line))
    except ValueError as e:
        errors.append((idx, line, str(e)))

print("有效记录:", valid_records)
print("无效记录:", errors)
```

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
        
    def is_passed(self):
        return self.score >= 60
        
    def __str__(self):
        return f"Student({self.name}, {self.score})"


students_objs = [Student(r["name"], r["score"]) for r in valid_records]
for s in students_objs:
    print(s, "及格" if s.is_passed() else "不及格")


import copy
orig = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(orig)
deep = copy.deepcopy(orig)
shallow["scores"].append(100)
print("原对象:", orig) # 浅拷贝导致原对象的嵌套列表也被改了
```
