```python
# ===== 欧几里得算法 =====

def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return abs(m)

print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300)  =", gcd(1200, 300))
print("gcd(5040, 60)   =", gcd(5040, 60))
```

```python
gcd(2018, 2019) == 1
gcd(1200, 300)  == 300
gcd(5040, 60)   == 60
```

```python
# ===== 测试整数属性 =====

def is_power_of_2(n):
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1


def is_power(b, n):
    if b < 1 or n < 1:
        return False
    if b == 1:
        return n == 1

    while n % b == 0:
        n //= b
    return n == 1
```

```python
print(is_power_of_2(1))   # True
print(is_power_of_2(8))   # True
print(is_power_of_2(12))  # False

print(is_power(3, 27))    # True
print(is_power(3, 30))    # False
print(is_power(1, 1))     # True
print(is_power(1, 2))     # False
```
